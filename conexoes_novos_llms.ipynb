```python
import openai
import os
from dotenv import load_dotenv

# Aponta explicitamente para o arquivo 'env.env'
load_dotenv("env.env")

chave = os.getenv('UNB_AK')

if not chave:
    print("ERRO: chave não foi carregada.")
else:
    client = openai.OpenAI(
        api_key=chave,
        base_url="https://ia-router.unb.br/"
    )
    
    print("Chave carregada. Testando conexão com o proxy da UnB.")
    
    response = client.chat.completions.create(
        model="UnB-Qwen/qwen3-8-27b",
        messages = [
            {
                "role": "user",
                "content": "O que é o mundo"
            }
        ]
    )
    
    print("\nResposta do Modelo:")
    print(response.choices[0].message.content)
```
