# Baixando Datasets

## Wikitext

### jsonl

```python
import json
from datasets import load_dataset

# 1. Download the Salesforce WikiText dataset from Hugging Face
# Options: "wikitext-103-raw-v1", "wikitext-103-v1", "wikitext-2-raw-v1", "wikitext-2-v1"
dataset_name = "Salesforce/wikitext"
dataset_config = "wikitext-103-raw-v1"

print(f"Downloading {dataset_name} ({dataset_config})...")
dataset = load_dataset(dataset_name, dataset_config)

# 2. Iterate through each split (train, test, validation) and save as JSONL
for split_name, split_data in dataset.items():
    output_filename = f"wikitext_{split_name}.jsonl"
    print(f"Saving {split_name} split to {output_filename}...")
    
    with open(output_filename, "w", encoding="utf-8") as f:
        for row in split_data:
            # Each row is a dictionary containing a 'text' key
            # Skip empty lines to keep the JSONL clean if desired
            if row["text"].strip():
                f.write(json.dumps(row, ensure_ascii=False) + "\n")

print("Download and conversion complete!")
```

## Congressional Bills

### jsonl

```python
import urllib.request
import os

# URLs originais do repositório
urls = {
    "train_bills.jsonl": "https://raw.githubusercontent.com/ahoho/topics/main/data/bills/train.metadata.jsonl",
    "test_bills.jsonl": "https://raw.githubusercontent.com/ahoho/topics/main/data/bills/test.metadata.jsonl"
}

# Pasta onde os arquivos serão salvos (neste caso, a pasta atual)
diretorio_atual = os.getcwd()

print("Iniciando o download...\n")

for nome_arquivo, url in urls.items():
    caminho_completo = os.path.join(diretorio_atual, nome_arquivo)
    print(f"Baixando: {nome_arquivo}")
    
    try:
        # Faz o download e salva o arquivo
        urllib.request.urlretrieve(url, caminho_completo)
        print(f"✅ Sucesso! Salvo em: {caminho_completo}\n")
    except Exception as e:
        print(f"❌ Erro ao baixar {nome_arquivo}: {e}\n")

print("Downloads concluídos!")
```

## Tweet Topic Single

### jsonl

```python
import jsonlines
from datasets import load_dataset

# 1. Load the dataset from Hugging Face
# You can specify the configuration name if needed (e.g., 'tweet_topic_single')
dataset = load_dataset("cardiffnlp/tweet_topic_single", "tweet_topic_single")

# 2. Iterate through each split (e.g., train_2020, test_2020, train_all, etc.)
for split_name, split_data in dataset.items():
    output_filename = f"{split_name}.jsonl"
    
    print(f"Saving {split_name} split to {output_filename}...")
    
    # 3. Write data row-by-row into a JSONL file
    with jsonlines.open(output_filename, mode='w') as writer:
        for row in split_data:
            writer.write(row)

print("Done! All splits have been downloaded and saved as JSONL.")
```
